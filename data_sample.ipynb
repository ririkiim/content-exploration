```python
# 정규화 함수
import re

def normalize_text(text):
    if not text:
        return ""

    text = str(text)

    # '저', '지음' 제거
    text = re.sub(r'\s*(저|지음)\s*$', '', text)

    # 공백 제거
    text = re.sub(r'\s+', '', text)

    # 특수문자 제거
    text = re.sub(r'[^\w가-힣]', '', text)

    return text.lower()
```

```python
books = [
    {"title": "칼의 노래", "author": "김훈"},
    {"title": "소년이 온다", "author": "한강"},
    {"title": "난장이가 쏘아올린 작은 공", "author": "조세희"},
    {"title": "토지", "author": "박경리"},
    {"title": "살인자의 기억법", "author": "김영하"},
    {"title": "아몬드", "author": "손원평"},
    {"title": "구운몽", "author": "김만중"},
    {"title": "완득이", "author": "김려령"},
    {"title": "82년생 김지영", "author": "조남주"},
    {"title": "지구 끝의 온실", "author": "김초엽"}
]
```

```python
import re

def is_same_work_title(target_title, result_title):
    target = normalize_text(target_title)
    result = normalize_text(result_title)

    # 1. 제목 완전 일치
    if result == target:
        return True

    # 2. 작품명 + 권수 형태
    # 예: 토지1, 토지2, 토지10
    if re.fullmatch(rf"{re.escape(target)}\d+", result):
        return True

    return False
```

```python
import requests
import pandas as pd

selected_books = []

for book in books:

    params = {
        "query": book["title"],
        "page": 1,
        "pageSize": 20
    }

    response = requests.get(
        url,
        params=params,
        headers=headers
    )

    data = response.json()
    items = data["data"]["items"]

    target_title = normalize_text(book["title"])
    target_author = normalize_text(book["author"])

    # 제목 + 저자가 일치하는 결과만 찾기
    matches = []

    for item in items:
        result_title = normalize_text(item.get("title", ""))
        result_author = normalize_text(item.get("author", ""))

        if (
            is_same_work_title(book["title"], item.get("title", ""))
            and target_author in result_author
        ):
            matches.append(item)

    # 여러 개면 첫 번째 선택
    if matches:
        selected = matches[0]

        selected_books.append({
            "title": book["title"],
            "author": book["author"],
            "isbn13": selected.get("isbn13"),
            "book_introduction": selected.get("contentDetail", {}).get("bookIntroduction")
        })

        print(f"선택: {book['title']} / {book['author']} "
              f"← YES24: {selected.get('title')}")

    else:
        print(f"❌ 찾지 못함: "
              f"{book['title']} / {book['author']}")
```

```python
# book_introduction 정제 (HTML 태그 제거, 공백 제거, 특수문자 제거)
from bs4 import BeautifulSoup
import re

def clean_text(text):
    if not text:
        return ""

    # HTML 태그 제거
    text = BeautifulSoup(text, "html.parser").get_text(" ", strip=True)

    # 연속된 공백 정리
    text = re.sub(r"\s+", " ", text)

    return text.strip()

books_df["book_introduction"] = books_df["book_introduction"].apply(clean_text)
books_df["book_introduction"]
```
